```python
import torch

from chapter4.gpt_config import GPT_CONFIG_124M
from chapter5.previous_chapters import generate_text_simple
from chapter5.train_model_simple import train_model_simple
from chapter5.text_encode import text_to_token_ids, token_ids_to_text

torch.manual_seed(123)
```

```python
import tiktoken
tokenizer = tiktoken.get_encoding("gpt2")
```

```python
file_path = "the_verdict.txt"
with open(file_path, "r", encoding="utf-8") as f:
    text_data = f.read()
train_ratio = 0.90
split_idx = int(train_ratio * len(text_data))
train_data = text_data[:split_idx]
validate_data = text_data[split_idx:]
```

```python
from previous_chapters import create_dataloader_v1, GPTModel
train_loader = create_dataloader_v1(
    train_data,
    batch_size=2,
    max_length=GPT_CONFIG_124M["context_length"],
    stride=GPT_CONFIG_124M["context_length"],
    drop_last=True,
    shuffle=True,
    num_workers=0
)

val_loader = create_dataloader_v1(
    validate_data,
    batch_size=2,
    max_length=GPT_CONFIG_124M["context_length"],
    stride=GPT_CONFIG_124M["context_length"],
    drop_last=False,
    shuffle=False,
    num_workers=0
)
```

```python
model = GPTModel(GPT_CONFIG_124M)
model.to(torch.device("cuda" if torch.cuda.is_available() else "cpu"))
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.0004, weight_decay=0.1
)
num_epochs = 10
train_losses, val_losses, tokens_seen = train_model_simple(
    model, train_loader, val_loader, optimizer, "cuda" if torch.cuda.is_available() else "cpu", num_epochs=num_epochs, eval_freq=5, eval_iter=5, start_context="Every effort moves you", tokenizer=tokenizer
)
```

```python
model.eval()
```

```python
token_ids = generate_text_simple(model=model, idx=text_to_token_ids("Every effort moves you", tokenizer), max_new_tokens=25, context_size=GPT_CONFIG_124M["context_length"])
print("output text: ", token_ids_to_text(token_ids, tokenizer))
```
