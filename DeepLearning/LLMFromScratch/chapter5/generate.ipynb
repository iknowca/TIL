```python
GPT_CONFIG_124M = {
    "vocab_size": 50257,
    "context_length": 256,
    "emb_dim": 768,
    "n_heads": 12,
    "n_layers": 12,
    "drop_rate": 0.1,
    "qkv_bias": False
}
```

```python
import tiktoken
tokenizer = tiktoken.get_encoding("gpt2")
```

```python
def text_to_token_ids(text, tokenizer):
    encoded = tokenizer.encode(text, allowed_special={"<|endoftext|>"} )
    encoded_tensor = torch.tensor(encoded).unsqueeze(0)
    return encoded_tensor
```

```python
def token_ids_to_text(token_ids, tokenizer):
    flat = token_ids.squeeze(0)
    return tokenizer.decode(flat.tolist())
```

```python
model = GPTModel(GPT_CONFIG_124M)
```

```python
from train_model_simple import generate
```

```python
token_ids = generate(model=model, idx=text_to_token_ids("Every effort moves you", tokenizer),  max_new_tokens=15, context_size=GPT_CONFIG_124M["context_length"], top_k=None, temperature=0)
```

```python
print("output: \n", token_ids_to_text(token_ids, tokenizer))
```
