```python
import numpy as np
```

```python
# 파티에 50명이 왔는데 그중 3명은 보스턴에 다녀왔다.
# 무작위로 3명을 선택했을때, 그 중 보스턴에 다녀온 사람이 없을 확률은?
nb = 0
N = 100000
```

```python
for i in range(N):
    s = np.random.randint(0, 50, 3)
    fail = False
    for t in range(3):
        if (s[t] < 4):
            fail = True
    if not fail:
        nb += 1
print("No Boston in the fall = %0.4f" %(nb / N))
```
