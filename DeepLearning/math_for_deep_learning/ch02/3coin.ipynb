```python
import numpy as np
```

```python
N = 10000000
M = 3
```

```python
heads = np.zeros(M+1)
```

```python
for i in range(N):
    flips = np.random.randint(0, 2, M)
    h, _ = np.bincount(flips, minlength = 2)
    heads[h] += 1
```

```python
prob = heads/N
print("Probabilities: %s" % np.array2string(prob))
```
