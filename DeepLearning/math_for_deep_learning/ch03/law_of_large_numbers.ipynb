```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = []
for n in np.linspace(1, 8, 30):
    t = np.random.normal(1, 1, size=int(10**n))
    m.append(t.mean())
```

```python
plt.plot(np.linspace(1, 8, 30), m)
plt.hlines(1, 1, 8,colors="black", linestyles="dashed", linewidth=0.5)
```
