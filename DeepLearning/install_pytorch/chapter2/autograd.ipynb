```python
import torch
```

```python
x = torch.tensor([1.], requires_grad=True)
print(x)
```

```python
x = torch.tensor([1.])
print(x)
print(x.requires_grad)

x.requires_grad = True
print(x)
print(x.requires_grad)
```

```python
x = torch.tensor([1.],  requires_grad=True)
y= x**2
print(y)

print(x.grad)
y.backward() # 미분, 단 requires_grad=True인 변수에 대해
print(x.grad)
```

```python
x = torch.tensor([1.],  requires_grad=True)
y= x**2
print(y)
# y.retain_grad()

z = 3 *y
print(z)

z.backward()
print(x.grad)
print(y.grad)
```

```python
x=torch.tensor([1.], requires_grad=True)
a = x** 2
b = a + 1
c = b**2
c.backward()
print(x.grad)
```

```python
x = torch.tensor([1.], requires_grad=True)
y = torch.tensor([1.], requires_grad=True)

z = 2*x**2 + y**2

print(z)
z.backward()
print(x.grad)
print(y.grad)
```

```python
x=torch.tensor([1., 2., 3.], requires_grad=True)
y = torch.sum(x**2)
y.backward()
print(y)
print(x.grad)
```

```python
x=torch.tensor([1.], requires_grad=True)
x.requires_grad=False
y=x**2
print(y)
```

```python
x=torch.tensor([2.], requires_grad=True)
x2 = x.detach()
print(x)
print(x2)

y = x**2
y2 = x2**2

print(y)
print(y2)
```

```python
x = torch.tensor([1. ], requires_grad=True)
with torch.no_grad():
    y = x**2
    print(x.requires_grad)
    print(y)

y = x**2
print(y)
x = torch.tensor([1.], requires_grad=True)
x.requires_grad = False
y = x**2
print(x.requires_grad)
print(x.grad_fn)
```
