```python
import sys
sys.path.append('..')
from util import most_similar, create_co_matrix, ppmi
from dataset import ptb
```

```python
window_size = 2
wordvec_size = 100
```

```python
corpus, word_to_id, id_to_word = ptb.load_data('train')
vocab_size = len(word_to_id)
print("co matrix calculation")
co_matrix = create_co_matrix(corpus, vocab_size, window_size)
print('ppmi calculation')
ppmi = ppmi(co_matrix)
```

```python
print('SVD calculation')
from sklearn.utils.extmath import randomized_svd
U, S, V = randomized_svd(ppmi, n_components=wordvec_size, n_iter=5, random_state=None)
word_vecs = U[:, :wordvec_size]
```

```python
querys = ['you', 'year', 'car', 'toyota']
for query in querys:
    most_similar(query, word_to_id, id_to_word, word_vecs, top=5)
```
