```python
text = "You say goodbye and I say hello."
```

```python
text = text.lower()
text = text.replace('.', ' .')
text
```

```python
words = text.split(' ')
words
```

```python
word_to_id = {}
id_to_word = {}

for word in words:
    if word not in word_to_id:
        new_id = len(word_to_id)
        word_to_id[word] = new_id
        id_to_word[new_id] = word
```

```python
print(id_to_word)
print(word_to_id)
```

```python
import numpy as np
```

```python
corpus = [word_to_id[word] for word in words]
corpus = np.array(corpus)
corpus
```
